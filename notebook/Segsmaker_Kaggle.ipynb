```python
# @title Bootstrap - run once per session
import importlib
import os
import shutil
import subprocess
import sys
from pathlib import Path

RAW_BASE = 'https://raw.githubusercontent.com/N3iKos/segsmaker-prallel/main'
WORKDIR = Path('/kaggle/working') if Path('/kaggle').exists() else Path.cwd()
WORKDIR.mkdir(parents=True, exist_ok=True)

print('Installing notebook widget support...')
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'ipywidgets', 'jupyterlab_widgets'], check=False)

helper_path = WORKDIR / 'segsmaker_widgets.py'
local_helper = Path.cwd() / 'script' / 'KC' / 'widgets.py'
if local_helper.exists():
    shutil.copy2(local_helper, helper_path)
else:
    url = f'{RAW_BASE}/script/KC/widgets.py'
    result = subprocess.run(['curl', '-fLo', str(helper_path), url], capture_output=True, text=True)
    if result.returncode != 0:
        print(f'Widget helper download failed:\n{result.stderr}')
        raise SystemExit(1)

if str(WORKDIR) not in sys.path:
    sys.path.insert(0, str(WORKDIR))

import segsmaker_widgets as segsw
importlib.reload(segsw)
print('Ready. Run the installer cell next.')
```

## WebUI Installer
Pilih WebUI dan masukkan token lewat widget Kaggle. Tidak perlu edit variabel Python.

```python
# @title WebUI Installer
import subprocess
import sys
from pathlib import Path

_setup_py = WORKDIR / 'setup.py'
_url = f'{RAW_BASE}/script/KC/setup.py'
_result = subprocess.run(['curl', '-fLo', str(_setup_py), _url], capture_output=True, text=True)
if _result.returncode != 0:
    print(f'Setup script download failed:\n{_result.stderr}')
    raise SystemExit(1)

segsw.display_installer(_setup_py)
```
